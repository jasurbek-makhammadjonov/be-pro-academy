```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python

path = kagglehub.dataset_download("grandeurkoe/boston-house-price-data")

print("Path to dataset files:", path)

csv_path = Path(path) / "boston.csv"


df = pd.read_csv(csv_path)

print("Kaggle data is ready")
print(df.head())
```

```python
df.info()
```

```python
# 1. Histogram of PRICE
plt.figure(figsize=(8,5))
sns.histplot(df["PRICE"], bins=30, kde=False, color="skyblue")
plt.title("Distribution of House Prices")
plt.xlabel("Price ($1000s)")
plt.ylabel("Frequency")
plt.show()

# 2. Bar chart of CHAS (categorical column)
plt.figure(figsize=(6,4))
sns.countplot(x="CHAS", data=df, palette="Set2")
plt.title("Count of Houses by Charles River Proximity")
plt.xlabel("CHAS (1 = near river, 0 = not)")
plt.ylabel("Count")
plt.show()

# 3. Correlation heatmap
plt.figure(figsize=(12,8))
corr = df.corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Boston Housing Dataset")
plt.xlabel("Features")
plt.ylabel("Features")
plt.show()
```
